```python
# Week 5 Project: Analytics Reports and Insights Documentation

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Week 5 Project Started Successfully!")
print("Analytics Reports and Insights Documentation")
```

```python
# Creating a sample retail sales dataset

data = {
    "Product": [
        "Laptop", "Phone", "Headphones",
        "Laptop", "Phone", "Headphones",
        "Laptop", "Phone", "Headphones"
    ],
    "Region": [
        "North", "North", "North",
        "South", "South", "South",
        "West", "West", "West"
    ],
    "Units_Sold": [
        12, 30, 25,
        9, 24, 20,
        7, 18, 15
    ],
    "Revenue": [
        720000, 450000, 50000,
        540000, 360000, 40000,
        420000, 270000, 30000
    ]
}

df = pd.DataFrame(data)

print("Sample Retail Sales Dataset")
display(df)
```

```python
# Step 4: Basic Data Analysis

# Calculate total revenue
total_revenue = df["Revenue"].sum()

# Calculate total units sold
total_units = df["Units_Sold"].sum()

# Calculate product-wise revenue
product_revenue = df.groupby("Product")["Revenue"].sum()

# Calculate region-wise revenue
region_revenue = df.groupby("Region")["Revenue"].sum()

print("WEEK 5 - DATA ANALYSIS REPORT")
print("--------------------------------")
print("Total Revenue: Rs.", total_revenue)
print("Total Units Sold:", total_units)

print("\nProduct-wise Revenue:")
print(product_revenue)

print("\nRegion-wise Revenue:")
print(region_revenue)
```

```python
# Step 5: Product-wise Revenue Visualization

plt.figure(figsize=(8, 5))

product_revenue.plot(kind="bar")

plt.title("Product-wise Revenue Analysis")
plt.xlabel("Product")
plt.ylabel("Revenue (Rs.)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# Step 6: Region-wise Revenue Visualization

plt.figure(figsize=(8, 5))

region_revenue.plot(kind="bar")

plt.title("Region-wise Revenue Analysis")
plt.xlabel("Region")
plt.ylabel("Revenue (Rs.)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# Step 7: Business Insights

best_product = product_revenue.idxmax()
lowest_product = product_revenue.idxmin()

best_region = region_revenue.idxmax()
lowest_region = region_revenue.idxmin()

print("WEEK 5 - BUSINESS INSIGHTS")
print("----------------------------")

print("1. Highest Revenue Product:", best_product)
print("2. Lowest Revenue Product:", lowest_product)
print("3. Highest Revenue Region:", best_region)
print("4. Lowest Revenue Region:", lowest_region)

print("\nBusiness Findings:")
print("- Laptops generate the highest total revenue.")
print("- Headphones generate the lowest total revenue.")
print("- North region has the highest revenue.")
print("- West region has the lowest revenue.")

print("\nNote: Findings are based on an illustrative sample dataset.")
```
